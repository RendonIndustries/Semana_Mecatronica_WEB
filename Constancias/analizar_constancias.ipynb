```python
# Extraer texto de una página específica (ejemplo: primera página de conferencias)
def extraer_texto_pagina(pdf_path, numero_pagina=0):
    """Extrae el texto de una página específica del PDF"""
    try:
        with open(pdf_path, 'rb') as file:
            pdf_reader = pypdf.PdfReader(file)
            if 0 <= numero_pagina < len(pdf_reader.pages):
                page = pdf_reader.pages[numero_pagina]
                texto = page.extract_text()
                return texto
            else:
                return f"Error: La página {numero_pagina} no existe. El PDF tiene {len(pdf_reader.pages)} páginas."
    except Exception as e:
        return f"Error al leer el PDF: {e}"

# Ejemplo: Extraer texto de la primera página de conferencias
if conferencias_path.exists():
    print("Texto de la primera página de constancias de conferencias:")
    print("-" * 60)
    texto = extraer_texto_pagina(conferencias_path, 0)
    print(texto[:500] + "..." if len(texto) > 500 else texto)
```

## Generación de Códigos QR para Validación

Esta sección agrega códigos QR únicos a cada constancia para validación y seguridad.

```python
# Instalar librerías necesarias para QR y edición de PDFs
import subprocess
import sys

def install_package(package):
    try:
        __import__(package)
    except ImportError:
        print(f"Instalando {package}...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", package, "--quiet"])

# Instalar qrcode y pdf-lib (para Python usamos qrcode y reportlab o pdf-lib equivalente)
install_package("qrcode[pil]")
install_package("Pillow")

print("✓ Librerías para QR instaladas correctamente")
```

```python
# Importar librerías para QR
import qrcode
from PIL import Image
import hashlib
from datetime import datetime

# URL base para validación (ajustar según el dominio)
BASE_URL = "https://semanameca.upiiz.ipn.mx"  # Cambiar por tu dominio
# BASE_URL = "http://localhost:3000"  # Para pruebas locales

def generar_hash_constancia(nombre, actividad, tipo, fecha=None):
    """Genera un hash único para la constancia"""
    if fecha is None:
        fecha = datetime.now().isoformat()
    
    # Crear string único
    datos = f"{nombre}|{actividad}|{tipo}|{fecha}"
    # Generar hash SHA256
    hash_obj = hashlib.sha256(datos.encode('utf-8'))
    hash_hex = hash_obj.hexdigest()[:16]  # Primeros 16 caracteres
    return hash_hex.upper()

def generar_qr_constancia(nombre, actividad, tipo, hash_constancia):
    """Genera un código QR para la constancia"""
    # Crear URL de validación
    url_validacion = f"{BASE_URL}/api/constancias/validar?hash={hash_constancia}"
    
    # Datos del QR (formato JSON)
    datos_qr = {
        "tipo": "constancia",
        "evento": "Semana de Mecatrónica 2025",
        "nombre": nombre,
        "actividad": actividad,
        "tipo_constancia": tipo,
        "hash": hash_constancia,
        "url_validacion": url_validacion
    }
    
    # Crear QR
    qr = qrcode.QRCode(
        version=1,
        error_correction=qrcode.constants.ERROR_CORRECT_M,
        box_size=10,
        border=2,
    )
    qr.add_data(url_validacion)  # URL para validación rápida
    qr.make(fit=True)
    
    # Crear imagen del QR
    img_qr = qr.make_image(fill_color="black", back_color="white")
    
    # Redimensionar a tamaño adecuado para PDF (aprox 150x150 píxeles)
    img_qr = img_qr.resize((150, 150), Image.Resampling.LANCZOS)
    
    return img_qr, hash_constancia, url_validacion

# Probar generación de QR
nombre_test = "María Guadalupe Fernández Domínguez"
actividad_test = "Conferencias Generales"
tipo_test = "conferencias"

img_qr_test, hash_test, url_test = generar_qr_constancia(nombre_test, actividad_test, tipo_test, None)
print(f"✓ QR generado correctamente")
print(f"  Hash: {hash_test}")
print(f"  URL: {url_test}")
print(f"  Tamaño imagen: {img_qr_test.size}")
```

## Extracción de Constancias Individuales

Esta sección extrae cada página del PDF como un archivo PDF individual, nombrado según el participante.

```python
# Función para limpiar y normalizar nombres de archivo
import re

def limpiar_nombre_archivo(nombre):
    """Limpia el nombre para usarlo como nombre de archivo"""
    if not nombre:
        return ""
    # Remover caracteres especiales y espacios múltiples
    nombre = re.sub(r'[<>:"/\\|?*]', '', nombre)
    nombre = re.sub(r'\s+', ' ', nombre)
    nombre = nombre.strip()
    return nombre

def extraer_nombre_participante(texto_pagina):
    """Extrae el nombre del participante del texto de la página"""
    if not texto_pagina:
        return None
    # El nombre generalmente está en la primera línea
    lineas = texto_pagina.strip().split('\n')
    # Filtrar líneas vacías
    lineas = [l.strip() for l in lineas if l.strip()]
    
    if lineas:
        # La primera línea suele ser el nombre
        nombre = lineas[0].strip()
        # Remover espacios extra y caracteres no deseados
        nombre = re.sub(r'\s+', ' ', nombre)
        # Validar que no sea muy corto (probablemente no es un nombre)
        if len(nombre) < 3:
            return None
        return nombre
    return None

def extraer_nombre_conferencia_taller(texto_pagina):
    """Extrae el nombre de la conferencia o taller del texto de la página"""
    if not texto_pagina:
        return None
    # El nombre de la conferencia/taller generalmente está en la segunda línea
    lineas = texto_pagina.strip().split('\n')
    # Filtrar líneas vacías
    lineas = [l.strip() for l in lineas if l.strip()]
    
    if len(lineas) >= 2:
        # La segunda línea suele ser el nombre de la conferencia/taller
        nombre = lineas[1].strip()
        # Remover espacios extra
        nombre = re.sub(r'\s+', ' ', nombre)
        # Validar que no sea muy corto
        if len(nombre) < 3:
            return None
        return nombre
    return None

# Probar la función con varias páginas de ejemplo
if conferencias_path.exists():
    print("Probando extracción de nombres con las primeras 5 páginas:")
    print("-" * 60)
    with open(conferencias_path, 'rb') as file:
        pdf_reader = pypdf.PdfReader(file)
        for i in range(min(5, len(pdf_reader.pages))):
            pagina = pdf_reader.pages[i]
            texto = pagina.extract_text()
            nombre = extraer_nombre_participante(texto)
            conferencia = extraer_nombre_conferencia_taller(texto)
            nombre_participante = limpiar_nombre_archivo(nombre) if nombre else "Sin_nombre"
            nombre_conf = limpiar_nombre_archivo(conferencia) if conferencia else "Sin_conferencia"
            nombre_archivo = f"{nombre_participante} - {nombre_conf}" if conferencia else nombre_participante
            print(f"Página {i+1}: {nombre} - {conferencia} -> {nombre_archivo}.pdf")
```

```python
# Función para extraer todas las páginas como PDFs individuales
def extraer_pdfs_individuales(pdf_path, carpeta_destino, tipo="conferencias"):
    """
    Extrae cada página del PDF como un archivo PDF individual.
    Usa el nombre del participante y el nombre de la conferencia/taller en el nombre del archivo.
    
    Args:
        pdf_path: Ruta al PDF original
        carpeta_destino: Carpeta donde se guardarán los PDFs individuales
        tipo: Tipo de constancia (para logging)
    """
    # Crear carpeta de destino si no existe
    carpeta_destino.mkdir(parents=True, exist_ok=True)
    
    pdfs_creados = []
    pdfs_con_errores = []
    
    try:
        with open(pdf_path, 'rb') as file:
            pdf_reader = pypdf.PdfReader(file)
            total_pages = len(pdf_reader.pages)
            
            print(f"\n{'='*60}")
            print(f"Extrayendo {total_pages} constancias de {tipo}...")
            print(f"{'='*60}")
            
            for i, page in enumerate(pdf_reader.pages):
                try:
                    # Extraer texto para obtener el nombre y la conferencia/taller
                    texto = page.extract_text()
                    nombre = extraer_nombre_participante(texto)
                    nombre_actividad = extraer_nombre_conferencia_taller(texto)
                    
                    # Si no se puede extraer el nombre, usar un nombre genérico
                    if not nombre or len(nombre.strip()) == 0:
                        nombre = f"Constancia_{i+1:04d}"
                    
                    # Limpiar nombres para archivo
                    nombre_participante = limpiar_nombre_archivo(nombre)
                    nombre_actividad_limpio = limpiar_nombre_archivo(nombre_actividad) if nombre_actividad else ""
                    
                    # Construir nombre del archivo
                    if nombre_actividad_limpio:
                        nombre_archivo = f"{nombre_participante} - {nombre_actividad_limpio}"
                    else:
                        nombre_archivo = nombre_participante
                    
                    # Limitar longitud total del nombre (Windows tiene límite de 260 caracteres)
                    if len(nombre_archivo) > 200:
                        # Si es muy largo, truncar primero el nombre de la actividad
                        if nombre_actividad_limpio:
                            max_nombre = len(nombre_participante) + 3  # +3 para " - "
                            max_actividad = 200 - max_nombre
                            nombre_actividad_limpio = nombre_actividad_limpio[:max_actividad]
                            nombre_archivo = f"{nombre_participante} - {nombre_actividad_limpio}"
                        else:
                            nombre_archivo = nombre_participante[:200]
                    
                    # Asegurar que el nombre no esté vacío
                    if not nombre_archivo or nombre_archivo.strip() == "":
                        nombre_archivo = f"Constancia_{i+1:04d}"
                    
                    # Crear PDF individual
                    pdf_writer = pypdf.PdfWriter()
                    pdf_writer.add_page(page)
                    
                    # Ruta del archivo de salida
                    archivo_salida = carpeta_destino / f"{nombre_archivo}.pdf"
                    
                    # Si el archivo ya existe, es porque hay duplicados (mismo nombre y misma actividad)
                    # En este caso, agregar un número al final
                    contador = 1
                    archivo_original = archivo_salida
                    while archivo_salida.exists():
                        nombre_base = archivo_original.stem
                        archivo_salida = carpeta_destino / f"{nombre_base}_{contador}.pdf"
                        contador += 1
                    
                    # Guardar PDF individual
                    with open(archivo_salida, 'wb') as output_file:
                        pdf_writer.write(output_file)
                    
                    pdfs_creados.append({
                        'pagina': i + 1,
                        'nombre': nombre,
                        'actividad': nombre_actividad,
                        'archivo': archivo_salida.name
                    })
                    
                    # Mostrar progreso cada 50 páginas
                    if (i + 1) % 50 == 0:
                        print(f"  ✓ Procesadas {i + 1}/{total_pages} páginas...")
                        
                except Exception as e:
                    print(f"  ✗ Error en página {i+1}: {e}")
                    pdfs_con_errores.append({
                        'pagina': i + 1,
                        'error': str(e)
                    })
            
            print(f"\n{'='*60}")
            print(f"✓ Extracción completada!")
            print(f"  - PDFs creados: {len(pdfs_creados)}")
            print(f"  - Errores: {len(pdfs_con_errores)}")
            print(f"  - Ubicación: {carpeta_destino.absolute()}")
            print(f"{'='*60}")
            
            return pdfs_creados, pdfs_con_errores
            
    except Exception as e:
        print(f"❌ Error al procesar el PDF: {e}")
        import traceback
        traceback.print_exc()
        return [], []

print("✅ Función definida correctamente")
```

### Verificar PDFs Extraídos

```python
# Verificar PDFs extraídos
def verificar_pdfs_extraidos(carpeta):
    """Verifica los PDFs extraídos en una carpeta"""
    if not carpeta.exists():
        print(f"❌ La carpeta {carpeta} no existe")
        return
    
    pdfs = list(carpeta.glob("*.pdf"))
    print(f"\n📁 Carpeta: {carpeta}")
    print(f"📄 Total de PDFs encontrados: {len(pdfs)}")
    
    if len(pdfs) > 0:
        print(f"\nPrimeros 10 archivos:")
        for pdf in sorted(pdfs)[:10]:
            tamaño = pdf.stat().st_size / 1024  # KB
            print(f"  - {pdf.name} ({tamaño:.2f} KB)")
        
        if len(pdfs) > 10:
            print(f"  ... y {len(pdfs) - 10} archivos más")
    
    return len(pdfs)

# Verificar carpetas
print("Verificación de PDFs extraídos:")
print("="*60)

carpeta_conf = Path("Conferencias") / "Individuales"
num_conf = verificar_pdfs_extraidos(carpeta_conf)

carpeta_tall = Path("Talleres") / "Individuales"
num_tall = verificar_pdfs_extraidos(carpeta_tall)

if num_conf and num_tall:
    print(f"\n📊 Resumen:")
    print(f"  - Conferencias: {num_conf} PDFs")
    print(f"  - Talleres: {num_tall} PDFs")
    print(f"  - Total: {num_conf + num_tall} PDFs")
```

```python
# Función para buscar texto específico en todas las páginas
def buscar_texto_en_pdf(pdf_path, texto_buscar):
    """Busca un texto específico en todas las páginas del PDF"""
    resultados = []
    try:
        with open(pdf_path, 'rb') as file:
            pdf_reader = pypdf.PdfReader(file)
            
            for i, page in enumerate(pdf_reader.pages):
                texto = page.extract_text()
                if texto_buscar.lower() in texto.lower():
                    resultados.append({
                        'pagina': i + 1,
                        'texto_encontrado': texto[:300]  # Primeros 300 caracteres
                    })
            
            return resultados
    except Exception as e:
        print(f"Error: {e}")
        return []

# Ejemplo de uso (descomentar para usar)
# texto_a_buscar = "ejemplo"
# if conferencias_path.exists():
#     resultados = buscar_texto_en_pdf(conferencias_path, texto_a_buscar)
#     print(f"Se encontró '{texto_a_buscar}' en {len(resultados)} páginas")
#     for resultado in resultados[:5]:  # Mostrar solo los primeros 5
#         print(f"\nPágina {resultado['pagina']}:")
#         print(resultado['texto_encontrado'][:200])
```
